# 4. Descrevam as informações contidas nos dados

### 4.1. Significado dos registros antes de calcular estatísticas.


- **O que uma linha representa?**
  1. Tabela de Acidentes (acidentes-de-transito-acidentes):
    - Uma linha representa: Um sinistro de trânsito individual ocorrido em uma determinada data, hora e local de Porto Alegre.
  2. Tabela de Vítimas (acidentes-de-transito-vitimas):
    - Uma linha representa: Uma pessoa envolvida com lesão/óbito em um acidente específico (uma mesma ocorrência com 3 feridos terá 3 linhas nesta tabela).
  3. Tabela de Sinalização Gráfica (sinalizacao-grafica):
    - Uma linha representa: Um elemento físico de sinalização (ex.: uma placa, pintura de faixa de pedestres, marcação de solo ou tachão) instalado em um ponto da cidade.

- **Como os dados foram produzidos?**
  1. Ocorrências de Trânsito e Vítimas: Originam-se do registro dos boletins de ocorrência lavrados pelos agentes de trânsito e equipes de emergência no local dos acidentes.
  2. Sinalização Gráfica: Provenientes do cadastro georreferenciado mantido pelo setor de Engenharia de Tráfego da EPTC à medida que novas sinalizações são projetadas, instaladas ou revitalizadas
- **Qual é a cobertura temporal e geográfica?**
  - Cobertura Geográfica:
    - Abrangência: Limites do município de Porto Alegre / RS.
    - Granularidade: Coordenadas exatas de Latitude e Longitude, cruzamentos, logradouros (ruas/avenidas) e bairros da cidade.
  - Cobertura Temporal:
    - Abrangência: Histórico plurianual completo (disponibilizado anualmente pela EPTC, cobrindo com precisão os anos de 2010 até 2025/2026).
    - Granularidade: Carimbo de data e hora exatos (Ano, Mês, Dia, Dia da Semana e Hora/Minuto da ocorrência do acidente).

- **Quem ou o que pode estar ausente da base?**
  1. Acidentes leves e sem feridos (podemos ter menos acidentes na base do que realmente aconteceu)
  2. Sinalizações informais, temporárias ou danificadas
  3. Direção da sinalização

- **Há mais de uma tabela? Como elas se relacionam?**
  1. Acidentes (acidentes-de-transito-acidentes) $\leftrightarrow$ Vítimas (acidentes-de-transito-vitimas):
    - Relacionamento: $1 : N$ (Um para Muitos)
    - Chave de Ligação: O campo **IDACIDENTE** (ou identificador único da ocorrência)
    - Como funciona: Um único acidente pode ter zero ou mais vítimas associadas. Cruzando essas duas tabelas pelo **IDACIDENTE**, conseguimos saber a gravidade real de cada sinistro (se houve feridos graves ou óbitos).
  2. Acidentes $\leftrightarrow$ Sinalização Gráfica (sinalizacao-grafica):
    - Relacionamento: Espacial / Geográfico (por Proximidade e Logradouro).
    - Chaves de Ligação: Coordenadas de Latitude/Longitude e/ou o nome da rua/avenida (Logradouro / Bairro)
    - Como funciona: Como a base de sinalização lista os elementos físicos (faixas, placas, pinturas) instalados pela cidade com suas coordenadas ou endereços, o cruzamento é feito verificando a distância/presença de sinalização no raio do ponto onde ocorreu o acidente.

## 5.2. Funções reutilizáveis

```python
def resumo_estatistico(df, nome_dataset="Dataset"):
    """
    Gera um panorama completo de dimensões, duplicatas, tipos de dados e nulos de um DataFrame.
    """
    print(f"📊 RESUMO: {nome_dataset.upper()}")
    print(f"• Linhas: {df.shape[0]:,} | Colunas: {df.shape[1]}".replace(',', '.'))
    print(f"• Linhas completamente duplicadas: {df.duplicated().sum()}")
    print("-" * 50)
    
    # Monta a tabela detalhada por coluna
    resumo = pd.DataFrame({
        'Tipo de Dado': df.dtypes.astype(str),
        'Valores Nulos': df.isnull().sum(),
        '% Nulos': (df.isnull().sum() / len(df) * 100).round(2),
        'Valores Únicos': df.nunique()
    })
    
    return resumo
```

```python
def analisar_distribuicao_dispersao(df, colunas, grupo=None, nome_dataset="Dataset"):
    """
    Aplica a análise de distribuição e dispersão para uma coluna ou uma lista de colunas.
    Valores de idade fora do intervalo plausível são ignorados apenas nos gráficos.
    """
    if isinstance(colunas, str):
        lista_colunas = [colunas]
    else:
        lista_colunas = colunas

    for col in lista_colunas:
        if col not in df.columns:
            print(f"⚠️ Aviso: A coluna '{col}' não foi encontrada em '{nome_dataset}' e será ignorada.\n")
            continue
        
        if not pd.api.types.is_numeric_dtype(df[col]):
            print(f"⚠️ Aviso: A coluna '{col}' não é numérica e será ignorada.\n")
            continue

        if grupo is None:
            dados = df[col].dropna()
            if col.lower() in {'idade', 'age'}:
                dados_originais = dados
                dados = dados[dados.between(0, 120)]
                descartados = len(dados_originais) - len(dados)
                if descartados:
                    print(f"⚠️ {descartados} valores de idade fora de 0 a 120 foram ignorados nos gráficos.")

            if dados.empty:
                print(f"⚠️ Aviso: A coluna '{col}' não possui valores válidos para análise.\n")
                continue

            media = dados.mean()
            mediana = dados.median()
            desvio_padrao = dados.std()
            q25, q75 = dados.quantile(0.25), dados.quantile(0.75)
            iqr = q75 - q25
            minimo, maximo = dados.min(), dados.max()
            assimetria = dados.skew()
            
            print(f"📊 DISTRIBUIÇÃO E DISPERSÃO: {col.upper()} ({nome_dataset})")
            print("-" * 65)
            print(f"• Registros válidos: {len(dados):,}".replace(',', '.'))
            print(f"• Média: {media:.2f} | Mediana: {mediana:.2f}")
            print(f"• Desvio Padrão: {desvio_padrao:.2f} | IQR: {iqr:.2f}")
            print(f"• Amplitude Total: Mín = {minimo:.2f} | Máx = {maximo:.2f}")
            print(f"• Assimetria (Skewness): {assimetria:.2f}")
            print("-" * 65)

            fig, axes = plt.subplots(1, 2, figsize=(12, 3.5))
            
            sns.histplot(dados, kde=True, ax=axes[0], color='skyblue', bins=20)
            axes[0].axvline(media, color='red', linestyle='--', label=f'Média ({media:.2f})')
            axes[0].axvline(mediana, color='green', linestyle='-', label=f'Mediana ({mediana:.2f})')
            axes[0].set_title(f'Histograma/KDE - {col}')
            axes[0].legend()
            
            sns.boxplot(x=dados, ax=axes[1], color='lightgreen')
            axes[1].axvline(media, color='red', linestyle='--', label=f'Média ({media:.2f})')
            axes[1].set_title(f'Boxplot - {col}')
            axes[1].legend()

            plt.tight_layout()
            plt.show()
            print("\n" + "="*70 + "\n")

        else:
            if grupo not in df.columns:
                print(f"⚠️ Aviso: A coluna de grupo '{grupo}' não foi encontrada no DataFrame.\n")
                break

            print(f"📊 DISTRIBUIÇÃO E DISPERSÃO: {col.upper()} AGRUPADO POR '{grupo.upper()}' ({nome_dataset})")
            print("-" * 65)
            
            resumo_grupo = df.groupby(grupo)[col].agg(
                qtd=('count'),
                media=('mean'),
                mediana=('median'),
                std=('std'),
                iqr=lambda x: x.quantile(0.75) - x.quantile(0.25),
                min=('min'),
                max=('max')
            ).reset_index()
            
            display(resumo_grupo)
            print("-" * 65)

            fig, axes = plt.subplots(1, 2, figsize=(14, 4))
            
            sns.histplot(data=df, x=col, hue=grupo, kde=True, ax=axes[0], element="step")
            axes[0].set_title(f'Distribuição de {col} por {grupo}')
            
            sns.boxplot(data=df, x=grupo, y=col, ax=axes[1], palette="Set2")
            axes[1].set_title(f'Dispersão de {col} por {grupo}')
            axes[1].tick_params(axis='x', rotation=45)

            plt.tight_layout()
            plt.show()
            print("\n" + "="*70 + "\n")
```

```python
def plotar_frequencia_cardinalidade(df, colunas, max_categorias=20, nome_dataset="Dataset"):
    """
    Plota gráficos de barras com contagens absolutas e percentuais para 
    uma coluna categórica ou uma lista de colunas.
    
    Parâmetros:
      - df: DataFrame contendo os dados.
      - colunas: String ou lista de strings com os nomes das colunas.
      - max_categorias: Limite máximo de categorias únicas para plotar (evita gráficos ilegíveis).
      - nome_dataset: Nome para exibição nas mensagens e títulos.
    """
    # Normalizar entrada: se receber string, converte para lista
    if isinstance(colunas, str):
        lista_colunas = [colunas]
    else:
        lista_colunas = colunas

    for col in lista_colunas:
        # 1. Validações
        if col not in df.columns:
            print(f"⚠️ Aviso: A coluna '{col}' não foi encontrada em '{nome_dataset}'.\n")
            continue
        
        qtd_unicos = df[col].nunique(dropna=True)
        
        if qtd_unicos > max_categorias:
            print(f"⚠️ Ppulando '{col}': Possui {qtd_unicos} valores únicos (limite definido: {max_categorias}).\n")
            continue

        # 2. Cálculo das Frequências Absoluta e Relativa (%)
        contagem = df[col].value_counts(dropna=False).reset_index()
        contagem.columns = [col, 'Quantidade']
        contagem['Percentual (%)'] = (contagem['Quantidade'] / len(df) * 100).round(2)

        # Tratar nulos no rótulo para exibição clara
        contagem[col] = contagem[col].fillna('Nulo / Não Informado').astype(str)

        print(f"📊 FREQUÊNCIA E CARDINALIDADE: {col.upper()} ({nome_dataset})")
        print("-" * 60)
        display(contagem)
        print("-" * 60)

        # 3. Construção do Gráfico de Barras
        plt.figure(figsize=(10, 5))
        
        # Gráfico principal de barras
        ax = sns.barplot(
            data=contagem, 
            x='Quantidade', 
            y=col, 
            palette='Blues_r'
        )
        
        # Adicionar rótulos com quantidade e % ao lado de cada barra
        total = len(df)
        for p in ax.patches:
            largura = p.get_width()
            percentual = (largura / total) * 100
            ax.annotate(
                f'{int(largura):,} ({percentual:.1f}%)'.replace(',', '.'),
                (largura, p.get_y() + p.get_height() / 2.),
                ha='left', va='center',
                xytext=(5, 0),
                textcoords='offset points',
                fontsize=9,
                color='black'
            )

        plt.title(f'Distribuição de Categorias: {col} ({nome_dataset})', fontsize=12)
        plt.xlabel('Quantidade de Registros')
        plt.ylabel(col)
        plt.grid(axis='x', linestyle='--', alpha=0.6)
        
        # Ajustar limite do eixo X para os rótulos de texto caberem na tela
        plt.xlim(0, contagem['Quantidade'].max() * 1.25)
        
        plt.tight_layout()
        plt.show()
        print("\n" + "="*70 + "\n")
```

```python
def analisar_completude(df, nome):
    tabela = pd.DataFrame({
        "Nulos": df.isna().sum(),
        "Percentual de nulos": (df.isna().mean() * 100).round(2),
        "Valores preenchidos": df.notna().sum()
    }).sort_values("Percentual de nulos", ascending=False)

    print(f"{nome}: {df.shape[0]} linhas e {df.shape[1]} colunas")
    display(tabela)

    # Para facilitar a leitura, o gráfico mostra somente atributos com nulos.
    tabela_nulos = tabela[tabela["Nulos"] > 0].copy()

    if tabela_nulos.empty:
        print("Não há valores nulos neste conjunto de dados.")
        return

    tabela_nulos = tabela_nulos.rename_axis("Atributo").reset_index()
    largura = min(max(7, len(tabela_nulos) * 0.8), 14)
    fig, ax = plt.subplots(figsize=(largura, 5))

    sns.barplot(
        data=tabela_nulos,
        x="Atributo",
        y="Percentual de nulos",
        color="#d95f59",
        ax=ax
    )

    ax.set_ylim(0, 100)
    ax.set_xlabel("")
    ax.set_ylabel("Percentual de valores nulos (%)")
    ax.set_title(
        f"Percentual de valores nulos por atributo - {nome}",
        loc="left",
        pad=10,
        weight="bold"
    )
    ax.grid(axis="y", linestyle="--", alpha=0.35)
    ax.set_axisbelow(True)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)
    plt.xticks(rotation=45, ha="right")

    for barra, percentual, quantidade in zip(
        ax.patches,
        tabela_nulos["Percentual de nulos"],
        tabela_nulos["Nulos"]
    ):
        ax.text(
            barra.get_x() + barra.get_width() / 2,
            percentual + 2,
            f"{percentual:.1f}%\n({int(quantidade):,})".replace(",", "."),
            ha="center",
            va="bottom",
            fontsize=8,
            color="#8f302b"
        )

    plt.tight_layout()
    plt.show()
```

```python
def validar_coordenadas_ausentes(
    df,
    nome_dataset,
    colunas=("latitude", "longitude")
):
    dados_ausentes = pd.DataFrame({
        "Coordenada": list(colunas),
        "Quantidade": [df[coluna].isna().sum() for coluna in colunas],
        "Percentual (%)": [
            df[coluna].isna().mean() * 100
            for coluna in colunas
        ]
    })

    dados_ausentes["Percentual (%)"] = (
        dados_ausentes["Percentual (%)"].round(2)
    )

    altura = max(4.5, 2.5 + len(colunas) * 1.2)

    fig, ax = plt.subplots(figsize=(10, altura))

    sns.barplot(
        data=dados_ausentes,
        x="Coordenada",
        y="Percentual (%)",
        color="#d95f59",
        ax=ax
    )

    # Escala percentual completa: 0% a 100%
    ax.set_ylim(0, 100)

    ax.set_title(
        f"Valores ausentes de coordenadas — {nome_dataset}",
        fontsize=18,
        weight="bold",
        pad=15
    )
    ax.set_xlabel("Coordenada", fontsize=14)
    ax.set_ylabel("Percentual de valores ausentes (%)", fontsize=14)
    ax.tick_params(axis="both", labelsize=13)
    ax.grid(axis="y", linestyle="--", alpha=0.5)
    ax.set_axisbelow(True)

    for barra, quantidade, percentual in zip(
        ax.patches,
        dados_ausentes["Quantidade"],
        dados_ausentes["Percentual (%)"]
    ):
        ax.annotate(
            f"{quantidade:,} ({percentual:.2f}%)".replace(",", "."),
            (
                barra.get_x() + barra.get_width() / 2,
                barra.get_height()
            ),
            ha="center",
            va="bottom",
            fontsize=12,
            weight="bold",
            xytext=(0, 6),
            textcoords="offset points"
        )

    plt.tight_layout()
    plt.show()

    print(f"Validação de coordenadas ausentes: {nome_dataset}")
    print(f"Total de linhas analisadas: {len(df):,}".replace(",", "."))
    display(dados_ausentes)
```

```python
def analisar_categorias(
    df,
    nome_dataset,
    colunas_categoricas=None,
    max_categorias=20,
    max_colunas=12
):
    """Analisa frequências de categorias sem depender do Jinja2."""

    if colunas_categoricas is None:
        colunas_categoricas = df.select_dtypes(
            include=["object", "category"]
        ).columns.tolist()

    colunas_validas = [
        coluna
        for coluna in colunas_categoricas
        if coluna in df.columns
        and df[coluna].nunique(dropna=True) <= max_categorias
    ][:max_colunas]

    if not colunas_validas:
        print(f"Nenhuma coluna categórica adequada encontrada em {nome_dataset}.")
        return

    n_colunas = 1
    n_linhas = len(colunas_validas)

    fig, eixos = plt.subplots(
        n_linhas,
        n_colunas,
        figsize=(24, 7 * n_linhas),
        squeeze=False
    )

    eixos = eixos.flatten()

    for eixo, coluna in zip(eixos, colunas_validas):
        contagem = (
            df[coluna]
            .fillna("Não informado")
            .astype(str)
            .value_counts()
            .sort_values()
        )

        eixo.barh(
            contagem.index,
            contagem.values,
            color=sns.color_palette("Set2", len(contagem))
        )

        limite_x = contagem.max() * 1.25
        eixo.set_xlim(0, limite_x)
        eixo.set_title(coluna, fontsize=24, weight="bold")
        eixo.set_xlabel("Quantidade de registros", fontsize=19)
        eixo.tick_params(axis="both", labelsize=17)
        eixo.grid(axis="x", linestyle="--", alpha=0.5)

        for indice, valor in enumerate(contagem.values):
            percentual = valor / len(df) * 100
            eixo.text(
                valor + limite_x * 0.01,
                indice,
                f"{valor:,} ({percentual:.1f}%)".replace(",", "."),
                va="center",
                fontsize=13
            )

        tabela = pd.DataFrame({
            "Categoria": contagem.index,
            "Quantidade": contagem.values,
            "Percentual (%)": (
                contagem.values / len(df) * 100
            ).round(2)
        })

        print(f"\n{coluna} — {nome_dataset}")
        display(tabela)

    for eixo in eixos[len(colunas_validas):]:
        eixo.remove()

    fig.suptitle(
        f"Distribuição das categorias — {nome_dataset}",
        fontsize=24,
        weight="bold"
    )

    plt.tight_layout(rect=[0, 0, 1, 0.97])
    plt.show()
```

```python
def normalizar_dia(valor):
    if pd.isna(valor):
        return np.nan

    texto = str(valor).strip().lower()
    texto = unicodedata.normalize("NFKD", texto)
    texto = "".join(
        caractere
        for caractere in texto
        if not unicodedata.combining(caractere)
    )

    return mapa_dias.get(texto, texto)
```

```python
def analisar_dados_espaciais(
    df,
    nome_dataset,
    coluna_gravidade=None,
    latitude_limites=(-30.30, -29.90),
    longitude_limites=(-51.30, -51.00)
):
    dados = df.copy()

    # Converter coordenadas para números
    for coluna in ["latitude", "longitude"]:
        dados[coluna] = pd.to_numeric(
            dados[coluna]
            .astype(str)
            .str.replace(",", ".", regex=False),
            errors="coerce"
        )

    coordenadas_completas = dados[["latitude", "longitude"]].notna().all(axis=1)

    coordenadas_validas = (
        coordenadas_completas
        & dados["latitude"].between(*latitude_limites)
        & dados["longitude"].between(*longitude_limites)
    )

    resumo = pd.DataFrame({
        "Situação": [
            "Coordenadas completas",
            "Coordenadas ausentes ou não numéricas",
            "Coordenadas fora da faixa geográfica"
        ],
        "Quantidade": [
            coordenadas_completas.sum(),
            (~coordenadas_completas).sum(),
            (coordenadas_completas & ~coordenadas_validas).sum()
        ]
    })

    print(f"Análise espacial — {nome_dataset}")
    display(resumo)

    dados = dados[coordenadas_validas].copy()

    # Gráfico com classificação opcional
    plt.figure(figsize=(10, 8))

    if coluna_gravidade is not None:
        dados["acidente_grave"] = (
            dados["feridos_gr"].fillna(0).gt(0)
            | dados["mortes"].fillna(0).gt(0)
        )

        sns.scatterplot(
            data=dados,
            x="longitude",
            y="latitude",
            hue="acidente_grave",
            palette={False: "#74add1", True: "#d73027"},
            alpha=0.5,
            s=25
        )
        plt.legend(title="Acidente grave")
    else:
        sns.scatterplot(
            data=dados,
            x="longitude",
            y="latitude",
            color="#756bb1",
            alpha=0.5,
            s=25
        )

    plt.title(f"Distribuição espacial — {nome_dataset}")
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.tight_layout()
    plt.show()

    print(f"Registros utilizados no gráfico: {len(dados)}")

    return dados, resumo
```

```python
def analisar_nulos_por_ano(df, nome_dataset, coluna_data, apenas_colunas=None):
    dados = df.copy()

    dados[coluna_data] = pd.to_datetime(
        dados[coluna_data],
        errors="coerce"
    )

    dados["ano"] = dados[coluna_data].dt.year
    dados = dados.dropna(subset=["ano"]).copy()
    dados["ano"] = dados["ano"].astype(int)

    colunas = apenas_colunas or [
        coluna for coluna in df.columns
        if coluna != coluna_data
    ]

    colunas = [coluna for coluna in colunas if coluna in dados.columns]

    nulos_por_ano = (
        dados.groupby("ano")[colunas]
        .apply(lambda grupo: grupo.isna().sum())
    )

    print(f"Quantidade de valores nulos por ano — {nome_dataset}")
    display(nulos_por_ano)

    quantidade_anos = len(nulos_por_ano.columns)
    quantidade_colunas = len(nulos_por_ano.index)

    fig, ax = plt.subplots(
        figsize=(
            max(12, quantidade_anos * 1.5),
            max(8, quantidade_colunas * 0.45)
        )
    )

    sns.heatmap(
        nulos_por_ano.T,
        annot=True,
        fmt="g",
        cmap="Reds",
        linewidths=0.5,
        linecolor="white",
        annot_kws={"fontsize": 8},
        cbar_kws={"label": "Quantidade de nulos"},
        ax=ax
    )

    ax.set_title(
        f"Quantidade de valores nulos por ano — {nome_dataset}",
        pad=15,
        weight="bold"
    )
    ax.set_xlabel("Ano")
    ax.set_ylabel("Atributo")
    ax.tick_params(axis="x", rotation=0)
    ax.tick_params(axis="y", rotation=0)

    plt.tight_layout()
    plt.show()
```

```python
def filtrar_coordenadas(df):
    df = df.copy()

    if {"latitude", "longitude"}.issubset(df.columns):
        df = df[
            df["latitude"].between(-90, 90)
            & df["longitude"].between(-180, 180)
        ]

    return df
```

## 5.5. Transformações e Limpeza de Dados

```python
# Preservar os dados brutos
acidentes = df_acidentes.copy()
vitimas = df_vitimas.copy()
sinalizacao = df_sinalizacao.copy()
```

```python
# Padronização
acidentes = padronizar_colunas(acidentes)
vitimas = padronizar_colunas(vitimas)
sinalizacao = padronizar_colunas(sinalizacao)
```

```python
# Limpeza
acidentes = padronizar_nulos(acidentes)
vitimas = padronizar_nulos(vitimas)
sinalizacao = padronizar_nulos(sinalizacao)
```

```python
# Converter datas
for df in [acidentes, vitimas]:
    for coluna in ["data", "data_extracao"]:
        if coluna in df.columns:
            df[coluna] = pd.to_datetime(df[coluna], errors="coerce")

if "implantacao" in sinalizacao.columns:
    sinalizacao["implantacao"] = pd.to_datetime(sinalizacao["implantacao"], errors="coerce")
```

```python
# Converter coordenadas e números
for df in [acidentes, vitimas, sinalizacao]:
    for coluna in ["latitude", "longitude"]:
        if coluna in df.columns:
            df[coluna] = pd.to_numeric(
                df[coluna]
                .astype("string")
                .str.replace(",", ".", regex=False),
                errors="coerce"
            )

for df in [acidentes, vitimas]:
    for coluna in colunas_numericas:
        if coluna in df.columns:
            df[coluna] = pd.to_numeric(
                df[coluna],
                errors="coerce"
            )
```

```python
# Remover duplicatas completas
acidentes = acidentes.drop_duplicates().copy()
vitimas = vitimas.drop_duplicates().copy()
sinalizacao = sinalizacao.drop_duplicates().copy()
```

```python
# Remover coordenadas impossíveis
acidentes = filtrar_coordenadas(acidentes)
vitimas = filtrar_coordenadas(vitimas)
sinalizacao = filtrar_coordenadas(sinalizacao)
```

```python
# Remover idades impossíveis, mantendo os nulos
if "idade" in vitimas.columns:
    vitimas.loc[
        ~vitimas["idade"].between(0, 120),
        "idade"
    ] = np.nan
```

```python
# Remover contagens negativas, mantendo os nulos
for coluna in ["feridos", "feridos_gr", "mortes", "morte_post", "fatais"]:
    if coluna in acidentes.columns:
        acidentes.loc[
            acidentes[coluna] < 0,
            coluna
        ] = np.nan
```

```python
# Variáveis derivadas dos acidentes
acidentes["ano"] = acidentes["data"].dt.year
acidentes["mes"] = acidentes["data"].dt.month
acidentes["dia_mes"] = acidentes["data"].dt.day

acidentes["dia_sem_normalizado"] = (
    acidentes["dia_sem"].apply(normalizar_dia)
)

acidentes["fim_de_semana"] = (
    acidentes["dia_sem_normalizado"]
    .isin(["sábado", "domingo"])
)

acidentes["acidente_grave"] = (
    acidentes["feridos_gr"].fillna(0).gt(0)
    | acidentes["mortes"].fillna(0).gt(0)
)

acidentes["acidente_fatal"] = (
    acidentes["mortes"].fillna(0).gt(0)
    | acidentes["morte_post"].fillna(0).gt(0)
)

acidentes["tem_vitima"] = (
    acidentes["cont_vit"].fillna(0).gt(0)
)
```

```python
# Faixas horárias
if "hora" in acidentes.columns:
    acidentes["hora_num"] = pd.to_numeric(
        acidentes["hora"]
        .astype("string")
        .str.extract(r"(\d{1,2})")[0],
        errors="coerce"
    )

    acidentes["faixa_horaria"] = pd.cut(
        acidentes["hora_num"],
        bins=[-1, 6, 12, 18, 24],
        labels=[
            "Madrugada",
            "Manhã",
            "Tarde",
            "Noite"
        ],
        right=False
    )
```

```python
# Relatório final da limpeza
for nome, df in {
    "Acidentes": acidentes,
    "Vítimas": vitimas,
    "Sinalização": sinalizacao
}.items():
    print(f"\n{nome}")
    print(f"Registros: {len(df):,}".replace(",", "."))
    print(f"Duplicatas: {df.duplicated().sum()}")
    print(f"Coordenadas ausentes: {df[['latitude', 'longitude']].isna().any(axis=1).sum()}")
    display(
        df.isna()
        .sum()
        .sort_values(ascending=False)
        .head(10)
        .to_frame("Valores nulos")
    )
```

```python
# Resumo
print("Limpeza concluída.")
print(f"Acidentes: {acidentes.shape}")
print(f"Vítimas: {vitimas.shape}")
print(f"Sinalização: {sinalizacao.shape}")
```
