# 4. Descrevam as informações contidas nos dados

### 4.1. Significado dos registros antes de calcular estatísticas.


- **O que uma linha representa?**
  1. Tabela de Acidentes (acidentes-de-transito-acidentes):
    - Uma linha representa: Um sinistro de trânsito individual ocorrido em uma determinada data, hora e local de Porto Alegre.
  2. Tabela de Vítimas (acidentes-de-transito-vitimas):
    - Uma linha representa: Uma pessoa envolvida com lesão/óbito em um acidente específico (uma mesma ocorrência com 3 feridos terá 3 linhas nesta tabela).
  3. Tabela de Sinalização Gráfica (sinalizacao-grafica):
    - Uma linha representa: Um elemento físico de sinalização (ex.: uma placa, pintura de faixa de pedestres, marcação de solo ou tachão) instalado em um ponto da cidade.

- **Como os dados foram produzidos?**
  1. Ocorrências de Trânsito e Vítimas: Originam-se do registro dos boletins de ocorrência lavrados pelos agentes de trânsito e equipes de emergência no local dos acidentes.
  2. Sinalização Gráfica: Provenientes do cadastro georreferenciado mantido pelo setor de Engenharia de Tráfego da EPTC à medida que novas sinalizações são projetadas, instaladas ou revitalizadas
- **Qual é a cobertura temporal e geográfica?**
  - Cobertura Geográfica:
    - Abrangência: Limites do município de Porto Alegre / RS.
    - Granularidade: Coordenadas exatas de Latitude e Longitude, cruzamentos, logradouros (ruas/avenidas) e bairros da cidade.
  - Cobertura Temporal:
    - Abrangência: Histórico plurianual completo (disponibilizado anualmente pela EPTC, cobrindo com precisão os anos de 2010 até 2025/2026).
    - Granularidade: Carimbo de data e hora exatos (Ano, Mês, Dia, Dia da Semana e Hora/Minuto da ocorrência do acidente).

- **Quem ou o que pode estar ausente da base?**
  1. Acidentes leves e sem feridos (podemos ter menos acidentes na base do que realmente aconteceu)
  2. Sinalizações informais, temporárias ou danificadas
  3. Direção da sinalização

- **Há mais de uma tabela? Como elas se relacionam?**
  1. Acidentes (acidentes-de-transito-acidentes) $\leftrightarrow$ Vítimas (acidentes-de-transito-vitimas):
    - Relacionamento: $1 : N$ (Um para Muitos)
    - Chave de Ligação: O campo **IDACIDENTE** (ou identificador único da ocorrência)
    - Como funciona: Um único acidente pode ter zero ou mais vítimas associadas. Cruzando essas duas tabelas pelo **IDACIDENTE**, conseguimos saber a gravidade real de cada sinistro (se houve feridos graves ou óbitos).
  2. Acidentes $\leftrightarrow$ Sinalização Gráfica (sinalizacao-grafica):
    - Relacionamento: Espacial / Geográfico (por Proximidade e Logradouro).
    - Chaves de Ligação: Coordenadas de Latitude/Longitude e/ou o nome da rua/avenida (Logradouro / Bairro)
    - Como funciona: Como a base de sinalização lista os elementos físicos (faixas, placas, pinturas) instalados pela cidade com suas coordenadas ou endereços, o cruzamento é feito verificando a distância/presença de sinalização no raio do ponto onde ocorreu o acidente.

```python
def resumo_estatistico(df, nome_dataset="Dataset"):
    """
    Gera um panorama completo de dimensões, duplicatas, tipos de dados e nulos de um DataFrame.
    """
    print(f"📊 RESUMO: {nome_dataset.upper()}")
    print(f"• Linhas: {df.shape[0]:,} | Colunas: {df.shape[1]}".replace(',', '.'))
    print(f"• Linhas completamente duplicadas: {df.duplicated().sum()}")
    print("-" * 50)
    
    # Monta a tabela detalhada por coluna
    resumo = pd.DataFrame({
        'Tipo de Dado': df.dtypes.astype(str),
        'Valores Nulos': df.isnull().sum(),
        '% Nulos': (df.isnull().sum() / len(df) * 100).round(2),
        'Valores Únicos': df.nunique()
    })
    
    return resumo
```

```python
def analisar_distribuicao_dispersao(df, colunas, grupo=None, nome_dataset="Dataset"):
    """
    Aplica a análise de distribuição e dispersão para uma coluna ou uma lista de colunas.
    Valores de idade fora do intervalo plausível são ignorados apenas nos gráficos.
    """
    if isinstance(colunas, str):
        lista_colunas = [colunas]
    else:
        lista_colunas = colunas

    for col in lista_colunas:
        if col not in df.columns:
            print(f"⚠️ Aviso: A coluna '{col}' não foi encontrada em '{nome_dataset}' e será ignorada.\n")
            continue
        
        if not pd.api.types.is_numeric_dtype(df[col]):
            print(f"⚠️ Aviso: A coluna '{col}' não é numérica e será ignorada.\n")
            continue

        if grupo is None:
            dados = df[col].dropna()
            if col.lower() in {'idade', 'age'}:
                dados_originais = dados
                dados = dados[dados.between(0, 120)]
                descartados = len(dados_originais) - len(dados)
                if descartados:
                    print(f"⚠️ {descartados} valores de idade fora de 0 a 120 foram ignorados nos gráficos.")

            if dados.empty:
                print(f"⚠️ Aviso: A coluna '{col}' não possui valores válidos para análise.\n")
                continue

            media = dados.mean()
            mediana = dados.median()
            desvio_padrao = dados.std()
            q25, q75 = dados.quantile(0.25), dados.quantile(0.75)
            iqr = q75 - q25
            minimo, maximo = dados.min(), dados.max()
            assimetria = dados.skew()
            
            print(f"📊 DISTRIBUIÇÃO E DISPERSÃO: {col.upper()} ({nome_dataset})")
            print("-" * 65)
            print(f"• Registros válidos: {len(dados):,}".replace(',', '.'))
            print(f"• Média: {media:.2f} | Mediana: {mediana:.2f}")
            print(f"• Desvio Padrão: {desvio_padrao:.2f} | IQR: {iqr:.2f}")
            print(f"• Amplitude Total: Mín = {minimo:.2f} | Máx = {maximo:.2f}")
            print(f"• Assimetria (Skewness): {assimetria:.2f}")
            print("-" * 65)

            fig, axes = plt.subplots(1, 2, figsize=(12, 3.5))
            
            sns.histplot(dados, kde=True, ax=axes[0], color='skyblue', bins=20)
            axes[0].axvline(media, color='red', linestyle='--', label=f'Média ({media:.2f})')
            axes[0].axvline(mediana, color='green', linestyle='-', label=f'Mediana ({mediana:.2f})')
            axes[0].set_title(f'Histograma/KDE - {col}')
            axes[0].legend()
            
            sns.boxplot(x=dados, ax=axes[1], color='lightgreen')
            axes[1].axvline(media, color='red', linestyle='--', label=f'Média ({media:.2f})')
            axes[1].set_title(f'Boxplot - {col}')
            axes[1].legend()

            plt.tight_layout()
            plt.show()
            print("\n" + "="*70 + "\n")

        else:
            if grupo not in df.columns:
                print(f"⚠️ Aviso: A coluna de grupo '{grupo}' não foi encontrada no DataFrame.\n")
                break

            print(f"📊 DISTRIBUIÇÃO E DISPERSÃO: {col.upper()} AGRUPADO POR '{grupo.upper()}' ({nome_dataset})")
            print("-" * 65)
            
            resumo_grupo = df.groupby(grupo)[col].agg(
                qtd=('count'),
                media=('mean'),
                mediana=('median'),
                std=('std'),
                iqr=lambda x: x.quantile(0.75) - x.quantile(0.25),
                min=('min'),
                max=('max')
            ).reset_index()
            
            display(resumo_grupo)
            print("-" * 65)

            fig, axes = plt.subplots(1, 2, figsize=(14, 4))
            
            sns.histplot(data=df, x=col, hue=grupo, kde=True, ax=axes[0], element="step")
            axes[0].set_title(f'Distribuição de {col} por {grupo}')
            
            sns.boxplot(data=df, x=grupo, y=col, ax=axes[1], palette="Set2")
            axes[1].set_title(f'Dispersão de {col} por {grupo}')
            axes[1].tick_params(axis='x', rotation=45)

            plt.tight_layout()
            plt.show()
            print("\n" + "="*70 + "\n")
```

```python
def plotar_frequencia_cardinalidade(df, colunas, max_categorias=20, nome_dataset="Dataset"):
    """
    Plota gráficos de barras com contagens absolutas e percentuais para 
    uma coluna categórica ou uma lista de colunas.
    
    Parâmetros:
      - df: DataFrame contendo os dados.
      - colunas: String ou lista de strings com os nomes das colunas.
      - max_categorias: Limite máximo de categorias únicas para plotar (evita gráficos ilegíveis).
      - nome_dataset: Nome para exibição nas mensagens e títulos.
    """
    # Normalizar entrada: se receber string, converte para lista
    if isinstance(colunas, str):
        lista_colunas = [colunas]
    else:
        lista_colunas = colunas

    for col in lista_colunas:
        # 1. Validações
        if col not in df.columns:
            print(f"⚠️ Aviso: A coluna '{col}' não foi encontrada em '{nome_dataset}'.\n")
            continue
        
        qtd_unicos = df[col].nunique(dropna=True)
        
        if qtd_unicos > max_categorias:
            print(f"⚠️ Ppulando '{col}': Possui {qtd_unicos} valores únicos (limite definido: {max_categorias}).\n")
            continue

        # 2. Cálculo das Frequências Absoluta e Relativa (%)
        contagem = df[col].value_counts(dropna=False).reset_index()
        contagem.columns = [col, 'Quantidade']
        contagem['Percentual (%)'] = (contagem['Quantidade'] / len(df) * 100).round(2)

        # Tratar nulos no rótulo para exibição clara
        contagem[col] = contagem[col].fillna('Nulo / Não Informado').astype(str)

        print(f"📊 FREQUÊNCIA E CARDINALIDADE: {col.upper()} ({nome_dataset})")
        print("-" * 60)
        display(contagem)
        print("-" * 60)

        # 3. Construção do Gráfico de Barras
        plt.figure(figsize=(10, 5))
        
        # Gráfico principal de barras
        ax = sns.barplot(
            data=contagem, 
            x='Quantidade', 
            y=col, 
            palette='Blues_r'
        )
        
        # Adicionar rótulos com quantidade e % ao lado de cada barra
        total = len(df)
        for p in ax.patches:
            largura = p.get_width()
            percentual = (largura / total) * 100
            ax.annotate(
                f'{int(largura):,} ({percentual:.1f}%)'.replace(',', '.'),
                (largura, p.get_y() + p.get_height() / 2.),
                ha='left', va='center',
                xytext=(5, 0),
                textcoords='offset points',
                fontsize=9,
                color='black'
            )

        plt.title(f'Distribuição de Categorias: {col} ({nome_dataset})', fontsize=12)
        plt.xlabel('Quantidade de Registros')
        plt.ylabel(col)
        plt.grid(axis='x', linestyle='--', alpha=0.6)
        
        # Ajustar limite do eixo X para os rótulos de texto caberem na tela
        plt.xlim(0, contagem['Quantidade'].max() * 1.25)
        
        plt.tight_layout()
        plt.show()
        print("\n" + "="*70 + "\n")
```

```python
def analisar_completude(df, nome):
    tabela = pd.DataFrame({
        "Nulos": df.isna().sum(),
        "Percentual de nulos": (df.isna().mean() * 100).round(2),
        "Valores preenchidos": df.notna().sum()
    }).sort_values("Percentual de nulos", ascending=False)

    print(f"{nome}: {df.shape[0]} linhas e {df.shape[1]} colunas")
    display(tabela)

    # Para facilitar a leitura, o gráfico mostra somente atributos com nulos.
    tabela_nulos = tabela[tabela["Nulos"] > 0].copy()

    if tabela_nulos.empty:
        print("Não há valores nulos neste conjunto de dados.")
        return

    tabela_nulos = tabela_nulos.rename_axis("Atributo").reset_index()
    largura = min(max(7, len(tabela_nulos) * 0.8), 14)
    fig, ax = plt.subplots(figsize=(largura, 5))

    sns.barplot(
        data=tabela_nulos,
        x="Atributo",
        y="Percentual de nulos",
        color="#d95f59",
        ax=ax
    )

    ax.set_ylim(0, 100)
    ax.set_xlabel("")
    ax.set_ylabel("Percentual de valores nulos (%)")
    ax.set_title(
        f"Percentual de valores nulos por atributo - {nome}",
        loc="left",
        pad=10,
        weight="bold"
    )
    ax.grid(axis="y", linestyle="--", alpha=0.35)
    ax.set_axisbelow(True)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)
    plt.xticks(rotation=45, ha="right")

    for barra, percentual, quantidade in zip(
        ax.patches,
        tabela_nulos["Percentual de nulos"],
        tabela_nulos["Nulos"]
    ):
        ax.text(
            barra.get_x() + barra.get_width() / 2,
            percentual + 2,
            f"{percentual:.1f}%\n({int(quantidade):,})".replace(",", "."),
            ha="center",
            va="bottom",
            fontsize=8,
            color="#8f302b"
        )

    plt.tight_layout()
    plt.show()
```

```python
def validar_coordenadas_ausentes(
    df,
    nome_dataset,
    colunas=("latitude", "longitude")
):
    dados_ausentes = pd.DataFrame({
        "Coordenada": list(colunas),
        "Quantidade": [df[coluna].isna().sum() for coluna in colunas],
        "Percentual (%)": [
            df[coluna].isna().mean() * 100
            for coluna in colunas
        ]
    })

    dados_ausentes["Percentual (%)"] = (
        dados_ausentes["Percentual (%)"].round(2)
    )

    altura = max(4.5, 2.5 + len(colunas) * 1.2)

    fig, ax = plt.subplots(figsize=(10, altura))

    sns.barplot(
        data=dados_ausentes,
        x="Coordenada",
        y="Percentual (%)",
        color="#d95f59",
        ax=ax
    )

    # Escala percentual completa: 0% a 100%
    ax.set_ylim(0, 100)

    ax.set_title(
        f"Valores ausentes de coordenadas — {nome_dataset}",
        fontsize=18,
        weight="bold",
        pad=15
    )
    ax.set_xlabel("Coordenada", fontsize=14)
    ax.set_ylabel("Percentual de valores ausentes (%)", fontsize=14)
    ax.tick_params(axis="both", labelsize=13)
    ax.grid(axis="y", linestyle="--", alpha=0.5)
    ax.set_axisbelow(True)

    for barra, quantidade, percentual in zip(
        ax.patches,
        dados_ausentes["Quantidade"],
        dados_ausentes["Percentual (%)"]
    ):
        ax.annotate(
            f"{quantidade:,} ({percentual:.2f}%)".replace(",", "."),
            (
                barra.get_x() + barra.get_width() / 2,
                barra.get_height()
            ),
            ha="center",
            va="bottom",
            fontsize=12,
            weight="bold",
            xytext=(0, 6),
            textcoords="offset points"
        )

    plt.tight_layout()
    plt.show()

    print(f"Validação de coordenadas ausentes: {nome_dataset}")
    print(f"Total de linhas analisadas: {len(df):,}".replace(",", "."))
    display(dados_ausentes)
```

```python
def analisar_categorias(
    df,
    nome_dataset,
    colunas_categoricas=None,
    max_categorias=20,
    max_colunas=12
):
    """Analisa frequências de categorias sem depender do Jinja2."""

    if colunas_categoricas is None:
        colunas_categoricas = df.select_dtypes(
            include=["object", "category"]
        ).columns.tolist()

    colunas_validas = [
        coluna
        for coluna in colunas_categoricas
        if coluna in df.columns
        and df[coluna].nunique(dropna=True) <= max_categorias
    ][:max_colunas]

    if not colunas_validas:
        print(f"Nenhuma coluna categórica adequada encontrada em {nome_dataset}.")
        return

    n_colunas = 1
    n_linhas = len(colunas_validas)

    fig, eixos = plt.subplots(
        n_linhas,
        n_colunas,
        figsize=(24, 7 * n_linhas),
        squeeze=False
    )

    eixos = eixos.flatten()

    for eixo, coluna in zip(eixos, colunas_validas):
        contagem = (
            df[coluna]
            .fillna("Não informado")
            .astype(str)
            .value_counts()
            .sort_values()
        )

        eixo.barh(
            contagem.index,
            contagem.values,
            color=sns.color_palette("Set2", len(contagem))
        )

        limite_x = contagem.max() * 1.25
        eixo.set_xlim(0, limite_x)
        eixo.set_title(coluna, fontsize=24, weight="bold")
        eixo.set_xlabel("Quantidade de registros", fontsize=19)
        eixo.tick_params(axis="both", labelsize=17)
        eixo.grid(axis="x", linestyle="--", alpha=0.5)

        for indice, valor in enumerate(contagem.values):
            percentual = valor / len(df) * 100
            eixo.text(
                valor + limite_x * 0.01,
                indice,
                f"{valor:,} ({percentual:.1f}%)".replace(",", "."),
                va="center",
                fontsize=13
            )

        tabela = pd.DataFrame({
            "Categoria": contagem.index,
            "Quantidade": contagem.values,
            "Percentual (%)": (
                contagem.values / len(df) * 100
            ).round(2)
        })

        print(f"\n{coluna} — {nome_dataset}")
        display(tabela)

    for eixo in eixos[len(colunas_validas):]:
        eixo.remove()

    fig.suptitle(
        f"Distribuição das categorias — {nome_dataset}",
        fontsize=24,
        weight="bold"
    )

    plt.tight_layout(rect=[0, 0, 1, 0.97])
    plt.show()
```

```python
def normalizar_dia(valor):
    if pd.isna(valor):
        return np.nan

    texto = str(valor).strip().lower()
    texto = unicodedata.normalize("NFKD", texto)
    texto = "".join(
        caractere
        for caractere in texto
        if not unicodedata.combining(caractere)
    )

    return mapa_dias.get(texto, texto)
```

```python
def analisar_dados_espaciais(
    df,
    nome_dataset,
    coluna_gravidade=None,
    latitude_limites=(-30.30, -29.90),
    longitude_limites=(-51.30, -51.00)
):
    dados = df.copy()

    # Converter coordenadas para números
    for coluna in ["latitude", "longitude"]:
        dados[coluna] = pd.to_numeric(
            dados[coluna]
            .astype(str)
            .str.replace(",", ".", regex=False),
            errors="coerce"
        )

    coordenadas_completas = dados[["latitude", "longitude"]].notna().all(axis=1)

    coordenadas_validas = (
        coordenadas_completas
        & dados["latitude"].between(*latitude_limites)
        & dados["longitude"].between(*longitude_limites)
    )

    resumo = pd.DataFrame({
        "Situação": [
            "Coordenadas completas",
            "Coordenadas ausentes ou não numéricas",
            "Coordenadas fora da faixa geográfica"
        ],
        "Quantidade": [
            coordenadas_completas.sum(),
            (~coordenadas_completas).sum(),
            (coordenadas_completas & ~coordenadas_validas).sum()
        ]
    })

    print(f"Análise espacial — {nome_dataset}")
    display(resumo)

    dados = dados[coordenadas_validas].copy()

    # Gráfico com classificação opcional
    plt.figure(figsize=(10, 8))

    if coluna_gravidade is not None:
        dados["acidente_grave"] = (
            dados["feridos_gr"].fillna(0).gt(0)
            | dados["mortes"].fillna(0).gt(0)
        )

        sns.scatterplot(
            data=dados,
            x="longitude",
            y="latitude",
            hue="acidente_grave",
            palette={False: "#74add1", True: "#d73027"},
            alpha=0.5,
            s=25
        )
        plt.legend(title="Acidente grave")
    else:
        sns.scatterplot(
            data=dados,
            x="longitude",
            y="latitude",
            color="#756bb1",
            alpha=0.5,
            s=25
        )

    plt.title(f"Distribuição espacial — {nome_dataset}")
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.tight_layout()
    plt.show()

    print(f"Registros utilizados no gráfico: {len(dados)}")

    return dados, resumo
```

```python
def analisar_nulos_por_ano(df, nome_dataset, coluna_data, apenas_colunas=None):
    dados = df.copy()

    dados[coluna_data] = pd.to_datetime(
        dados[coluna_data],
        errors="coerce"
    )

    dados["ano"] = dados[coluna_data].dt.year
    dados = dados.dropna(subset=["ano"]).copy()
    dados["ano"] = dados["ano"].astype(int)

    colunas = apenas_colunas or [
        coluna for coluna in df.columns
        if coluna != coluna_data
    ]

    colunas = [coluna for coluna in colunas if coluna in dados.columns]

    nulos_por_ano = (
        dados.groupby("ano")[colunas]
        .apply(lambda grupo: grupo.isna().sum())
    )

    print(f"Quantidade de valores nulos por ano — {nome_dataset}")
    display(nulos_por_ano)

    quantidade_anos = len(nulos_por_ano.columns)
    quantidade_colunas = len(nulos_por_ano.index)

    fig, ax = plt.subplots(
        figsize=(
            max(12, quantidade_anos * 1.5),
            max(8, quantidade_colunas * 0.45)
        )
    )

    sns.heatmap(
        nulos_por_ano.T,
        annot=True,
        fmt="g",
        cmap="Reds",
        linewidths=0.5,
        linecolor="white",
        annot_kws={"fontsize": 8},
        cbar_kws={"label": "Quantidade de nulos"},
        ax=ax
    )

    ax.set_title(
        f"Quantidade de valores nulos por ano — {nome_dataset}",
        pad=15,
        weight="bold"
    )
    ax.set_xlabel("Ano")
    ax.set_ylabel("Atributo")
    ax.tick_params(axis="x", rotation=0)
    ax.tick_params(axis="y", rotation=0)

    plt.tight_layout()
    plt.show()
```

## 5.7. Streamlit

```python
import os
import shutil
import signal
import subprocess
import sys
import time
from pathlib import Path
from urllib.error import URLError
from urllib.request import urlopen

projeto = Path.cwd()
app = projeto / "app.py"
porta = 8501
url_app = f"http://localhost:{porta}"


def streamlit_esta_rodando(url):
    try:
        with urlopen(f"{url}/_stcore/health", timeout=1) as resposta:
            return resposta.status == 200
    except (URLError, TimeoutError, OSError):
        return False


def aguardar(condicao, segundos):
    limite = time.time() + segundos
    while time.time() < limite:
        if condicao():
            return True
        time.sleep(0.5)
    return condicao()


def pids_escutando(porta):
    # Só quem está escutando na porta (não os navegadores conectados a ela).
    if shutil.which("lsof"):
        resultado = subprocess.run(
            ["lsof", "-t", f"-iTCP:{porta}", "-sTCP:LISTEN"],
            capture_output=True, text=True,
        )
        return [int(pid) for pid in resultado.stdout.split()]
    if shutil.which("fuser"):
        resultado = subprocess.run(
            ["fuser", f"{porta}/tcp"], capture_output=True, text=True
        )
        return [int(pid) for pid in resultado.stdout.split()]
    return []


def encerrar_streamlit():
    # Processo iniciado por esta célula em uma execução anterior.
    if "processo" in globals() and processo is not None and processo.poll() is None:
        processo.terminate()
        try:
            processo.wait(timeout=10)
        except subprocess.TimeoutExpired:
            processo.kill()

    # Qualquer outro Streamlit na porta (ex.: kernel reiniciado).
    for pid in pids_escutando(porta):
        if pid != os.getpid():
            os.kill(pid, signal.SIGTERM)
    if not aguardar(lambda: not streamlit_esta_rodando(url_app), 10):
        for pid in pids_escutando(porta):
            os.kill(pid, signal.SIGKILL)
        aguardar(lambda: not streamlit_esta_rodando(url_app), 5)


print("Aplicativo existe:", app.exists())
print("Caminho:", app)

# Sempre reinicia: assim o Streamlit roda a versão mais recente do código.
if streamlit_esta_rodando(url_app):
    print("Encerrando o Streamlit anterior...")
encerrar_streamlit()

processo = subprocess.Popen([
    sys.executable,
    "-m",
    "streamlit",
    "run",
    str(app),
    "--server.address",
    "localhost",
    "--server.port",
    str(porta),
], cwd=str(projeto))

if aguardar(lambda: streamlit_esta_rodando(url_app), 30):
    print(f"Streamlit iniciado em {url_app}")
else:
    print(f"O Streamlit ainda não respondeu em {url_app}; confira o log acima.")
```

## 5.8. Interpretação

### Interpretação dos resultados

A análise mostra que os acidentes de trânsito não se distribuem de forma uniforme ao longo do tempo e do espaço. Os gráficos anuais permitem comparar a quantidade de ocorrências entre os anos analisados, enquanto o gráfico de vítimas fatais mostra como a severidade dos acidentes varia nesse período.

O heatmap de dia da semana e horário indica os períodos com maior concentração de acidentes. Essa informação pode apoiar ações de fiscalização e planejamento da mobilidade, especialmente nos horários e dias que apresentam maior número de ocorrências.

A distribuição mensal permite observar diferenças entre os meses, embora essas variações não devam ser interpretadas isoladamente. Fatores como volume de tráfego, condições climáticas, obras e mudanças na circulação podem influenciar os resultados.

O gráfico que compara dias úteis e fins de semana ajuda a identificar em quais desses períodos os acidentes são mais frequentes. Essa comparação descreve a distribuição das ocorrências, mas não permite afirmar que um período causa mais acidentes.

O mapa de calor evidencia a concentração espacial dos acidentes em determinados pontos de Porto Alegre. As áreas mais intensas podem ser consideradas locais prioritários para investigações posteriores, fiscalização e avaliação das condições viárias.

A classificação de acidente grave foi definida pela presença de feridos graves ou mortes. Portanto, os acidentes classificados como graves representam ocorrências com maior severidade registrada na base. Essa classificação facilita a comparação entre os locais, períodos e tipos de acidente.

#### Relação entre acidentes e sinalização (dashboard)

Na aba "Sinalização × Acidentes" do dashboard, cada acidente foi ligado aos sinais próximos, considerando apenas os sinais já implantados no ano do acidente. Os números abaixo usam raio de 100 m e todos os anos (2021 a 2026). Eles mudam com o raio, por isso o dashboard permite ajustá-lo.

- **Densidade de sinais:** a proporção de acidentes graves cai de 8,6% nos locais com até 9 sinais em 100 m para 4,7% nos locais com 54 ou mais. Depois de controlar tipo de acidente, região, horário e ano, a razão do grupo com mais sinais em relação ao grupo com menos é de 0,63 (intervalo de confiança de 95%: 0,57 a 0,70): a associação diminui, mas não desaparece. Em colisões a queda se repete (de 5,5% para 2,0%); em atropelamentos não aparece, e a proporção fica em torno de 25% em todos os grupos.
- **Categorias de sinal:** controlando densidade, tipo de acidente e região, Estacionamento (razão de 0,77) e Setas e orientação (0,84) se associam a menos acidentes graves. Transporte coletivo (1,17) e Velocidade e moderadores (1,11) se associam a mais, o que provavelmente reflete onde esses sinais são instalados, como corredores de ônibus e vias rápidas. Divisão de pista, Pare, Bicicletas e Outras placas não mostram diferença clara, e Travessia de pedestres fica no limite (0,93).
- **Combinações de sinais:** a proporção de graves cai de 9,4% quando nenhuma categoria está por perto para 4,6% quando as nove estão.
- **Distância ao sinal mais próximo:** a mediana é de 11 m, e 92% dos acidentes ocorrem a até 50 m de algum sinal. Como o cadastro é denso, a distância separa pouco os acidentes. A proporção de graves vai de 6,6% (menos de 5 m) a 9,2% (100 a 200 m), mas as faixas distantes têm poucos acidentes.
- **Horário:** 29,2% dos acidentes ocorrem à noite, e 12,4% deles são graves, contra 4,3% de dia. A queda da gravidade com mais sinais é parecida de dia e de noite, sem indício de que as marcações no leito da via importem mais à noite.
- **Vítimas:** 72% das vítimas estão em acidentes com uma só vítima. Pedestres têm mais vítimas em acidentes graves (26,3%) do que condutores (19,0%). Para pedestres, a densidade de sinais não muda esse padrão, e ter travessia por perto também não (26,3% com travessia e 26,4% sem). Para ciclistas, a diferença é de 15,1% com sinalização de bicicletas e 20,0% sem, sugestiva, mas com intervalos de confiança que se sobrepõem.
- **Antes e depois da implantação:** em locais com sinal novo, comparados a locais de controle, o número de acidentes não mostra diferença clara na maioria das categorias. Divisão de pista, Bicicletas e Outras placas têm pequenas reduções (razões de 0,91, 0,81 e 0,86, com janela de 12 meses). Para acidentes graves, o intervalo de confiança é largo demais para concluir.
- **Pontos críticos:** o ranking por prioridade (muitos acidentes graves e pouca sinalização) serve como ponto de partida para priorizar locais, mas não substitui a verificação em campo.

De modo geral, os resultados ajudam a identificar padrões temporais, espaciais e de gravidade dos acidentes e mostram associações entre a sinalização do entorno e a gravidade. Entretanto, a análise não permite concluir que a sinalização, isoladamente, causa ou evita acidentes. Sinais costumam ser instalados onde já havia problema, e a base não traz volume e velocidade do tráfego nem o porte da via. O cadastro de sinalização inclui apenas os sinais em vigor na data da extração, a gravidade da vítima é a do acidente em que ela estava e os intervalos de confiança do antes e depois tendem a ser estreitos demais.

Também é importante considerar que os dados correspondem aos registros disponibilizados pela EPTC e podem conter valores ausentes, diferenças na cobertura temporal e subnotificação de acidentes leves ou sem vítimas. Assim, os resultados devem ser utilizados como apoio exploratório à tomada de decisão, e não como prova definitiva de causalidade.

### 6.2.1. Quais conjuntos/tipos de sinalização gráfica estão associados a uma menor proporção de acidentes graves em Porto Alegre?

**Resposta:** nenhum tipo de sinal, sozinho, se associa de forma clara a menos acidentes graves. O que mais se associa a menos graves é haver **mais sinais e mais categorias diferentes por perto**. Entre as categorias, **Estacionamento** e **Setas e orientação** se associam a menos graves, e **Transporte coletivo** e **Velocidade e moderadores**, a mais.

- **Comparação entre locais com e sem cada categoria** (razão de Mantel-Haenszel, que compara só locais com a mesma densidade de sinais, o mesmo tipo de acidente e a mesma região): Estacionamento 0,77 (0,72 a 0,83); Setas e orientação 0,84 (0,79 a 0,89); Travessia de pedestres 0,93 (0,86 a 0,99) e Pare e retenção 0,93 (0,86 a 1,00), no limite; Divisão de pista 1,03, Outras placas 0,96 e Bicicletas 0,96, sem diferença clara; Transporte coletivo 1,17 (1,10 a 1,25) e Velocidade e moderadores 1,11 (1,05 a 1,17), com mais graves. Esses dois últimos provavelmente refletem onde os sinais são instalados, como corredores de ônibus e vias rápidas.
- **Conjuntos de sinais:** a proporção de graves é de 9,4% quando nenhuma categoria está por perto e de 4,6% quando as nove estão. A combinação mais comum (oito categorias, sem Bicicletas) tem 5,0% de graves em 7.927 acidentes, e a de apenas Transporte coletivo tem 13,0% em 1.412.
- **Antes e depois da implantação** (janela de 12 meses, contra locais de controle): o número de acidentes não muda de forma clara na maioria das categorias, com pequenas reduções em Divisão de pista (razão de 0,91), Outras placas (0,86) e Bicicletas (0,81). Para acidentes graves, o intervalo de confiança é largo demais para concluir.

**Limites:** como a pergunta compara conjuntos de sinais, a densidade e a variedade pesam mais que qualquer categoria isolada, e as duas andam juntas com o tipo de via (avenidas têm mais sinais). O controle por região e tipo de acidente reduz essa confusão, mas não a elimina.

### 6.2.2. Quais são as zonas de acidentes graves em Porto Alegre e a quanto tempo/distância eles estão do elemento de sinalização mais próximo?

**Resposta:** as áreas com mais acidentes graves são, em sua maioria, trechos de grandes avenidas, e a distância ao sinal mais próximo é curta para quase todos eles: separa pouco os acidentes, embora a proporção de graves aumente com a distância.

- **Zonas:** dividindo a cidade em áreas de 200 m, 2.135 têm ao menos um acidente grave e 281 têm cinco ou mais. As 20 áreas com mais graves são todas avenidas (48 das 50 primeiras). As áreas com mais graves são Av. Protásio Alves × R. Dr. Murtinho (17 graves em 116 acidentes), Av. Presidente Castelo Branco (17 em 252) e Av. Protásio Alves × R. São Domingos (15 em 209). Pela prioridade (muitos graves e pouca sinalização), a primeira é Av. Assis Brasil × Av. Bernardino Silveira de Amorim, com 13 graves e mediana de 5 sinais em 100 m. Entre as 20 primeiras pela prioridade, 4 são estradas, que têm pouca sinalização.
- **Distância:** a mediana é de 11 m (11,4 m para os graves e 10,6 m para os demais), e 92% dos acidentes ocorrem a até 50 m de algum sinal. Com sinal a até 50 m, 6,5% dos acidentes são graves (73.402 acidentes); com sinal a mais de 200 m, 10,2% (482 acidentes). A faixa acima de 500 m tem só 75 acidentes e não permite concluir.
- **Tempo:** a base não traz o tempo até o sinal, apenas a distância.

**Limites:** poucos acidentes ocorrem longe de sinais, então o contraste "perto" contra "longe" tem amostra pequena. O cadastro só traz os sinais em vigor, e um local sem sinal por perto pode ter perdido um sinal que existia.

### 6.2.3. Existe correlação entre a densidade de sinalização e a severidade dos acidentes em um determinado raio?

**Resposta:** sim, há associação entre mais sinais e menos acidentes graves, forte nas proporções e fraca na correlação individual (Spearman de −0,054, porque o acidente grave é um evento raro).

- **Proporções:** a proporção de graves cai de 8,6% nos locais com até 9 sinais em 100 m para 4,7% nos com 54 ou mais.
- **Razão ajustada:** controlando tipo de acidente, região, horário e ano, o grupo com mais sinais tem razão de 0,63 (0,57 a 0,70) em relação ao grupo com menos: a associação diminui em relação à razão sem controle (0,54), mas não desaparece.
- **Por região:** Leste cai de 8,9% para 4,8%, Norte de 6,3% para 3,8% e Sul de 11,2% para 5,9%. O Centro fica quase plano (de 4,7% para 4,4%), porque quase não tem locais com pouca sinalização.
- **Por tipo de acidente:** em colisões a queda se repete (de 5,5% para 2,0%); em atropelamentos não aparece, e a proporção fica em torno de 25%.

**Limites:** áreas centrais e avenidas têm mais sinais, menor velocidade e mais fluxo, e o dado não separa esses fatores. A região dá um controle parcial.

### 6.2.4. Quais faixas horárias e dias da semana concentram o maior número de acidentes com vítimas?

**Resposta:** os acidentes se concentram em dias úteis e nos horários de pico, com destaque para o fim da tarde. À noite há menos acidentes, mas eles são mais graves.

- **Horário:** 18h concentra 7,9% dos acidentes, seguido de 17h (7,3%) e 8h (6,9%). Os maiores picos são quarta-feira às 18h (1.102 acidentes), quarta-feira às 8h (1.095) e sexta-feira às 18h (1.087).
- **Dias:** sexta-feira tem 17,0% dos acidentes e domingo, 8,0%. Fim de semana soma 19,1%, embora seja 2 dos 7 dias.
- **Com vítimas:** o pico é às 18h (8,8% dos acidentes com vítimas), e os maiores números são terça, quarta e sexta às 18h. O fim de semana pesa mais: 23,1% dos acidentes com vítimas, contra 19,1% do total.
- **Dia e noite:** 29,2% dos acidentes ocorrem à noite, e 12,4% deles são graves, contra 4,3% de dia. Atropelamentos têm pico às 18h (446).

**Limites:** os totais por horário refletem também o volume de tráfego em cada período, que a base não traz. Sem esse volume, não dá para dizer se um horário é mais perigoso por veículo ou apenas mais movimentado.

### 6.2.5. Como se distribuem as vítimas, por papel (condutor, ocupante, pedestre) e perfil, entre acidentes graves e não graves, conforme a densidade e a presença de sinalização por perto?

**Resposta:** pedestres têm proporção maior de vítimas em acidentes graves que condutores, e para eles nem a densidade de sinais nem a travessia por perto mudam o padrão. Para condutores e ocupantes, mais sinais se associam a menos graves. Para ciclistas, a sinalização de bicicletas aparece com menos graves, mas com incerteza.

- **Papel da vítima** (a tabela de vítimas não traz gravidade individual, então a vítima herda a do acidente): 26,3% dos pedestres estão em acidentes graves, contra 19,0% dos condutores. Só nos acidentes com uma vítima (72% delas), em que a gravidade é a da própria pessoa, a proporção em acidentes graves cai de 18,2% para 14,0% nos condutores e de 10,9% para 6,9% nos ocupantes à medida que os sinais aumentam, e fica em torno de 22% a 26% nos pedestres.
- **Pedestre × Travessia de pedestres:** 26,3% com travessia por perto e 26,4% sem.
- **Ciclista × Sinalização de bicicletas:** 15,1% com sinalização por perto e 20,0% sem, diferença sugestiva, mas com intervalos de confiança que se sobrepõem (11,9 a 19,1 contra 17,6 a 22,7).

**Limites:** a gravidade herdada do acidente só é exata com uma vítima. Há poucos ciclistas (1.331 vítimas), e sinais de bicicleta costumam estar em vias com ciclofaixa, que têm outras características.
